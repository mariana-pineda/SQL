```
%sql
-- SQL script for data quality checks in the d_product table within the purgo_playground schema

-- Validate 'item_nbr' for non-null entries and list count & sample records with null values
WITH Null_Item_Nbr AS (
  SELECT * 
  FROM purgo_playground.d_product
  WHERE item_nbr IS NULL
)
SELECT COUNT(*) AS Null_Item_Nbr_Count FROM Null_Item_Nbr;
SELECT * FROM Null_Item_Nbr LIMIT 5;

-- Validate 'sellable_qty' for non-null entries and list count & sample records with null values
WITH Null_Sellable_Qty AS (
  SELECT * 
  FROM purgo_playground.d_product
  WHERE sellable_qty IS NULL
)
SELECT COUNT(*) AS Null_Sellable_Qty_Count FROM Null_Sellable_Qty;
SELECT * FROM Null_Sellable_Qty LIMIT 5;

-- Validate 'prod_exp_dt' for correct 'YYYYMMDD' format and list count & sample records with incorrect format
WITH Incorrect_Format_Prod_Exp_Dt AS (
  SELECT * 
  FROM purgo_playground.d_product
  WHERE NOT prod_exp_dt RLIKE "^[0-9]{8}$"
)
SELECT COUNT(*) AS Incorrect_Format_Count FROM Incorrect_Format_Prod_Exp_Dt;
SELECT * FROM Incorrect_Format_Prod_Exp_Dt LIMIT 5;
```
