```
%sql
/*
  SQL Code for Data Quality Checks on d_product Table
  This script performs validation checks for non-null values and correct date formats.
*/

-- Use CTE to retrieve initial rows from the d_product table for validation purposes
WITH product_data AS (
  SELECT 
    prod_id,
    item_nbr,
    sellable_qty,
    prod_exp_dt
  FROM purgo_playground.d_product
)

-- Validate 'item_nbr' is not null
SELECT COUNT(*) AS null_item_nbr_count FROM product_data WHERE item_nbr IS NULL;
SELECT * FROM product_data WHERE item_nbr IS NULL LIMIT 5;

-- Validate 'sellable_qty' is not null
SELECT COUNT(*) AS null_sellable_qty_count FROM product_data WHERE sellable_qty IS NULL;
SELECT * FROM product_data WHERE sellable_qty IS NULL LIMIT 5;

-- Validate 'prod_exp_dt' is in 'YYYYMMDD' date format
-- Checking for correct date format by length and numeric pattern match
SELECT COUNT(*) AS invalid_prod_exp_dt_count 
FROM product_data 
WHERE LENGTH(prod_exp_dt) != 8 OR NOT prod_exp_dt RLIKE '^[0-9]{8}$';

SELECT * FROM product_data 
WHERE LENGTH(prod_exp_dt) != 8 OR NOT prod_exp_dt RLIKE '^[0-9]{8}$' 
LIMIT 5;

/* Clean up operations: End of validations */

```
