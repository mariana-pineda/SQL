```
%py
# PySpark script to mask last 4 digits of invoice numbers in d_product_revenue_clone table

# # Initialize SparkSession
# from pyspark.sql import SparkSession
# spark = SparkSession.builder.getOrCreate()

from pyspark.sql.functions import col, expr, substring
from pyspark.sql.types import StructType, StructField, StringType, BigIntType, DoubleType, DateType

# Defining schema for data consistency
schema = StructType([
    StructField("product_id", BigIntType(), True),
    StructField("product_name", StringType(), True),
    StructField("product_type", StringType(), True),
    StructField("revenue", BigIntType(), True),
    StructField("country", StringType(), True),
    StructField("customer_id", StringType(), True),
    StructField("purchased_date", DateType(), True),
    StructField("invoice_date", DateType(), True),
    StructField("invoice_number", BigIntType(), True),
    StructField("is_returned", BigIntType(), True),
    StructField("customer_satisfaction_score", BigIntType(), True),
    StructField("product_details", StringType(), True),
    StructField("customer_first_purchased_date", DateType(), True),
    StructField("customer_first_product", StringType(), True),
    StructField("customer_first_revenue", DoubleType(), True)
])

# Load the data from the original table
try:
    df_original = spark.table("purgo_playground.d_product_revenue")

    # Mask last 4 digits of invoice_number
    df_masked = df_original.withColumn("invoice_number",
        expr("CONCAT(SUBSTRING(CAST(invoice_number AS STRING), 1, LENGTH(CAST(invoice_number AS STRING)) - 4), '****')")
    )

    # Write the masked data to the cloned table
    df_masked.write.mode("overwrite").saveAsTable("purgo_playground.d_product_revenue_clone")
    
except Exception as e:
    print(f"Error: {str(e)}")

# # Stop SparkSession
# spark.stop()
```
