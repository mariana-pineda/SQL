```
%sql
-- Validate and display data issues in the d_product table

-- Validate 'item_nbr' is not null and display 5 sample records where 'item_nbr' is null
WITH ItemNbrNullRecords AS (
  SELECT *
  FROM purgo_databricks.purgo_playground.d_product
  WHERE item_nbr IS NULL
)
SELECT COUNT(*) AS item_nbr_null_count FROM ItemNbrNullRecords;

SELECT * FROM ItemNbrNullRecords LIMIT 5;

-- Validate 'sellable_qty' is not null and display 5 sample records where 'sellable_qty' is null
WITH SellableQtyNullRecords AS (
  SELECT *
  FROM purgo_databricks.purgo_playground.d_product
  WHERE sellable_qty IS NULL
)
SELECT COUNT(*) AS sellable_qty_null_count FROM SellableQtyNullRecords;

SELECT * FROM SellableQtyNullRecords LIMIT 5;

-- Validate 'prod_exp_dt' is in 'YYYYMMDD' format and display 5 sample records where 'prod_exp_dt' is not in 'YYYYMMDD' format
WITH IncorrectProdExpDtFormat AS (
  SELECT *
  FROM purgo_databricks.purgo_playground.d_product
  WHERE prod_exp_dt NOT REGEXP '^[0-9]{8}$'
)
SELECT COUNT(*) AS incorrect_prod_exp_dt_count FROM IncorrectProdExpDtFormat;

SELECT * FROM IncorrectProdExpDtFormat LIMIT 5;
```
