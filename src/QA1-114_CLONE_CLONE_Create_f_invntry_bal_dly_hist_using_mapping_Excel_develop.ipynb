```
%sql
/*
Purpose: Load and transform data into purgo_playground.f_invntry_bal_dly_hist table based on source mappings.
*/

/* 
Section: Create f_invntry_bal_dly_hist Table 
Description: Create the target table with appropriate data types and constraints.
*/
CREATE TABLE IF NOT EXISTS purgo_playground.f_invntry_bal_dly_hist (
    co_key STRING NOT NULL,
    g_account STRING NOT NULL,
    g_assignment_type STRING NOT NULL,
    g_capture_dt_yyyymmdd TIMESTAMP NOT NULL,
    g_company_cd STRING NOT NULL,
    g_company_currency_cd STRING NOT NULL,
    g_dnsa_dt_yyyymmdd TIMESTAMP,
    g_expiration_dt_yyyymmdd TIMESTAMP,
    g_item_nbr STRING NOT NULL,
    g_location_cd STRING NOT NULL,
    g_location_status_cd STRING NOT NULL,
    g_lot_effective_dt_yyyymmdd TIMESTAMP,
    g_lot_nbr STRING NOT NULL,
    g_lot_status_cd STRING NOT NULL,
    g_plant_cd STRING NOT NULL,
    g_qty_allocated DECIMAL(10,2),
    g_qty_available DECIMAL(10,2),
    g_qty_financial_nettable DECIMAL(10,2),
    g_qty_in_transit DECIMAL(10,2),
    g_qty_mrp_nettable DECIMAL(10,2),
    g_qty_on_hand DECIMAL(10,2),
    g_qty_shippable DECIMAL(10,2),
    g_source_system_cd STRING NOT NULL,
    g_unit_cost_company_currency DECIMAL(10,2),
    plant_key STRING NOT NULL,
    plant_location_key STRING NOT NULL,
    prod_key STRING NOT NULL,
    prod_plant_key STRING NOT NULL,
    prod_plant_location_key STRING NOT NULL,
    prod_plant_lot_key STRING NOT NULL,
    prod_plant_lot_location_key STRING NOT NULL
) USING DELTA;

/* 
Section: Add Column Comments 
Description: Document valid values and descriptions for each column.
*/
ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN co_key COMMENT 'Concatenate source_system_cd and g_company_cd';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_account COMMENT 'Must be 100% Complete Alphanumeric';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_assignment_type COMMENT 'Default value [Account]';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_capture_dt_yyyymmdd COMMENT 'Current date in YYYYMMDD format';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_company_cd COMMENT 'Hardcoded as 1050';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_company_currency_cd COMMENT 'Hardcoded as EUR';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_dnsa_dt_yyyymmdd COMMENT 'Expiration Date or NULL';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_expiration_dt_yyyymmdd COMMENT 'Default value 99991231 if not represented';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_item_nbr COMMENT 'Must be unique and alphanumeric';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_location_cd COMMENT 'Alphanumeric, default to none if not represented';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_location_status_cd COMMENT 'Hardcoded as none';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_lot_effective_dt_yyyymmdd COMMENT 'Default value 19000101 if not represented';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_lot_nbr COMMENT 'AlphaNumeric, default to none if blank';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_lot_status_cd COMMENT 'Hardcoded as none';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_plant_cd COMMENT 'Hardcoded as CAGER';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_qty_allocated COMMENT 'Derived from reservation_entry.quantity_base, negated if negative';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_qty_available COMMENT 'Derived from item_qty_on_hand.qty_on_hand for specific locations';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_qty_financial_nettable COMMENT 'Derived from item_qty_on_hand.qty_on_hand for specific locations';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_qty_in_transit COMMENT 'Hardcoded to NULL';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_qty_mrp_nettable COMMENT 'Derived from item_qty_on_hand.qty_on_hand for specific or AH- locations';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_qty_on_hand COMMENT 'Directly from item_qty_on_hand.qty_on_hand';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_qty_shippable COMMENT 'Derived from g_qty_available for specific locations';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_source_system_cd COMMENT 'Hardcoded as nav_ger';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN g_unit_cost_company_currency COMMENT 'Derived from itemwopd.Standard_Cost';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN plant_key COMMENT 'Concatenate source_system_cd and g_plant_cd';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN plant_location_key COMMENT 'Concatenate source_system_cd, g_plant_cd, and g_location_cd';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN prod_key COMMENT 'Concatenate source_system_cd and g_item_nbr';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN prod_plant_key COMMENT 'Concatenate source_system_cd, g_item_nbr, and g_plant_cd';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN prod_plant_location_key COMMENT 'Concatenate source_system_cd, g_item_nbr, g_plant_cd, and g_location_cd';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN prod_plant_lot_key COMMENT 'Concatenate source_system_cd, g_item_nbr, g_plant_cd, and g_lot_nbr';

ALTER TABLE purgo_playground.f_invntry_bal_dly_hist 
    ALTER COLUMN prod_plant_lot_location_key COMMENT 'Concatenate source_system_cd, g_item_nbr, g_plant_cd, g_lot_nbr, and g_location_cd';

/* 
Section: Load Data into f_invntry_bal_dly_hist 
Description: Insert transformed and joined data from source tables into the target table.
*/
INSERT INTO purgo_playground.f_invntry_bal_dly_hist
SELECT
    CONCAT(ile.src_sys_cd, '1050') AS co_key, -- Concatenate source_system_cd and g_company_cd
    smd.profit_center AS g_account, -- From sap_master_data.profit_center
    'Account' AS g_assignment_type, -- Hardcoded as 'Account'
    current_timestamp() AS g_capture_dt_yyyymmdd, -- Current timestamp
    '1050' AS g_company_cd, -- Hardcoded as '1050'
    'EUR' AS g_company_currency_cd, -- Hardcoded as 'EUR'
    CAST(ile.Expiration_Date AS TIMESTAMP) AS g_dnsa_dt_yyyymmdd, -- Expiration_Date from itemledgerentrieswopd
    CASE 
        WHEN ile.Expiration_Date IS NULL THEN CAST('9999-12-31' AS TIMESTAMP) 
        ELSE CAST(ile.Expiration_Date AS TIMESTAMP) 
    END AS g_expiration_dt_yyyymmdd, -- Default to '99991231' if not present
    TRIM(ile.item_no_) AS g_item_nbr, -- Trimmed Item_No_
    TRIM(ile.location_code) AS g_location_cd, -- Trimmed Location_Code
    'none' AS g_location_status_cd, -- Hardcoded as 'none'
    CAST(ile.Posting_Date AS TIMESTAMP) AS g_lot_effective_dt_yyyymmdd, -- Posting_Date from itemledgerentrieswopd
    CASE 
        WHEN TRIM(ile.serial_no_) = '' OR TRIM(ile.lot_no_) = '' THEN 'none' 
        ELSE TRIM(ile.lot_no_) 
    END AS g_lot_nbr, -- Default to 'none' if serial_no_ or lot_no_ is blank
    'none' AS g_lot_status_cd, -- Hardcoded as 'none'
    'CAGER' AS g_plant_cd, -- Hardcoded as 'CAGER'
    CASE 
        WHEN re.quantity_base < 0 THEN -re.quantity_base 
        ELSE re.quantity_base 
    END AS g_qty_allocated, -- Derived from reservation_entry.quantity_base
    CASE 
        WHEN iqoh.location_code IN ('HAUPT', 'GERMERING', 'LIN', 'MATRIUM') THEN iqoh.qty_on_hand 
        ELSE NULL 
    END AS g_qty_available, -- Based on specific locations
    CASE 
        WHEN iqoh.location_code IN ('HAUPT', 'GERMERING', 'LIN', 'MATRIUM') THEN iqoh.qty_on_hand 
        ELSE NULL 
    END AS g_qty_financial_nettable, -- Same as g_qty_available
    NULL AS g_qty_in_transit, -- Hardcoded to NULL
    CASE 
        WHEN iqoh.location_code IN ('HAUPT','QS-WE','GERMERING','SERVICE','LIN','MATRIUM') 
             OR TRIM(iqoh.location_code) LIKE 'AH-%' THEN iqoh.qty_on_hand 
        ELSE NULL 
    END AS g_qty_mrp_nettable, -- Based on specific locations or starting with 'AH-'
    iqoh.qty_on_hand AS g_qty_on_hand, -- Direct from item_qty_on_hand.qty_on_hand
    CASE 
        WHEN iqoh.location_code = 'HAUPT' THEN iqoh.qty_on_hand 
        ELSE NULL 
    END AS g_qty_shippable, -- Based on location 'HAUPT'
    'nav_ger' AS g_source_system_cd, -- Hardcoded as 'nav_ger'
    iw.Standard_Cost AS g_unit_cost_company_currency, -- From itemwopd.Standard_Cost
    CONCAT(ile.src_sys_cd, 'CAGER') AS plant_key, -- Concatenate source_system_cd and g_plant_cd
    CONCAT(ile.src_sys_cd, 'CAGER', TRIM(ile.location_code)) AS plant_location_key, -- Concatenate source_system_cd, g_plant_cd, and g_location_cd
    CONCAT(ile.src_sys_cd, TRIM(ile.item_no_)) AS prod_key, -- Concatenate source_system_cd and g_item_nbr
    CONCAT(ile.src_sys_cd, TRIM(ile.item_no_), 'CAGER') AS prod_plant_key, -- Concatenate source_system_cd, g_item_nbr, and g_plant_cd
    CONCAT(ile.src_sys_cd, TRIM(ile.item_no_), 'CAGER', TRIM(ile.location_code)) AS prod_plant_location_key, -- Concatenate source_system_cd, g_item_nbr, g_plant_cd, and g_location_cd
    CONCAT(ile.src_sys_cd, TRIM(ile.item_no_), 'CAGER', 
           CASE WHEN TRIM(ile.lot_no_) = '' THEN 'none' ELSE TRIM(ile.lot_no_) END) AS prod_plant_lot_key, -- Concatenate with lot_no_ or 'none'
    CONCAT(ile.src_sys_cd, TRIM(ile.item_no_), 'CAGER', 
           CASE WHEN TRIM(ile.lot_no_) = '' THEN 'none' ELSE TRIM(ile.lot_no_) END, 
           TRIM(ile.location_code)) AS prod_plant_lot_location_key -- Concatenate with lot_no_ or 'none' and location_code
FROM purgo_playground.item_qty_on_hand iqoh
LEFT OUTER JOIN purgo_playground.itemledgerentrieswopd ile
    ON TRIM(ile.item_no_) = TRIM(iqoh.item_no_) 
    AND TRIM(ile.location_code) = TRIM(iqoh.location_code)
LEFT OUTER JOIN purgo_playground.sap_master_data smd
    ON ile.item_no_ = smd.item_no_
LEFT OUTER JOIN purgo_playground.reservation_entry re
    ON TRIM(ile.item_no_) = TRIM(re.item_no_) 
    AND TRIM(ile.location_code) = TRIM(re.location_code)
LEFT OUTER JOIN purgo_playground.itemwopd iw
    ON ile.item_no_ = iw.item_no_
WHERE
    ile.posting_date = (
        SELECT MAX(posting_date) 
        FROM purgo_playground.itemledgerentrieswopd 
        WHERE item_no_ = ile.item_no_ 
          AND location_code = ile.location_code
    )
    AND re.reservation_status = '0' 
    AND re.source_type = '37';

/* 
Section: Cluster and Optimize 
Description: Apply clustering keys and Z-ordering for performance.
*/
OPTIMIZE purgo_playground.f_invntry_bal_dly_hist
ZORDER BY (g_item_nbr, g_location_cd);

/* 
Section: Data Validation 
Description: Ensure that data has been loaded correctly.
*/
/* Validate no column mismatch */
SELECT COUNT(*) AS column_mismatch
FROM INFORMATION_SCHEMA.COLUMNS
WHERE table_schema = 'purgo_playground'
  AND table_name = 'f_invntry_bal_dly_hist'
  AND column_name NOT IN (
      'co_key', 'g_account', 'g_assignment_type', 'g_capture_dt_yyyymmdd', 
      'g_company_cd', 'g_company_currency_cd', 'g_dnsa_dt_yyyymmdd', 
      'g_expiration_dt_yyyymmdd', 'g_item_nbr', 'g_location_cd', 
      'g_location_status_cd', 'g_lot_effective_dt_yyyymmdd', 'g_lot_nbr', 
      'g_lot_status_cd', 'g_plant_cd', 'g_qty_allocated', 'g_qty_available', 
      'g_qty_financial_nettable', 'g_qty_in_transit', 'g_qty_mrp_nettable', 
      'g_qty_on_hand', 'g_qty_shippable', 'g_source_system_cd', 
      'g_unit_cost_company_currency', 'plant_key', 'plant_location_key', 
      'prod_key', 'prod_plant_key', 'prod_plant_location_key', 
      'prod_plant_lot_key', 'prod_plant_lot_location_key'
  );

/* Validate g_assignment_type is 'Account' */
SELECT
    COUNT(*) AS invalid_g_assignment_type
FROM purgo_playground.f_invntry_bal_dly_hist
WHERE g_assignment_type <> 'Account';

/* Validate g_company_cd is '1050' */
SELECT
    COUNT(*) AS invalid_g_company_cd
FROM purgo_playground.f_invntry_bal_dly_hist
WHERE g_company_cd <> '1050';

/* Validate g_source_system_cd is 'nav_ger' */
SELECT
    COUNT(*) AS invalid_g_source_system_cd
FROM purgo_playground.f_invntry_bal_dly_hist
WHERE g_source_system_cd <> 'nav_ger';

/* Validate completeness of g_account */
SELECT
    COUNT(*) AS incomplete_g_account
FROM purgo_playground.f_invntry_bal_dly_hist
WHERE g_account IS NULL OR g_account = '';

/* Validate g_capture_dt_yyyymmdd format */
SELECT
    COUNT(*) AS invalid_capture_dt
FROM purgo_playground.f_invntry_bal_dly_hist
WHERE DATE_FORMAT(g_capture_dt_yyyymmdd, 'yyyyMMdd') IS NULL;

/* Validate g_dnsa_dt_yyyymmdd is valid or default */
SELECT
    COUNT(*) AS invalid_dnsa_dt
FROM purgo_playground.f_invntry_bal_dly_hist
WHERE (g_dnsa_dt_yyyymmdd NOT BETWEEN CAST('0001-01-01' AS TIMESTAMP) AND CAST('9999-12-31' AS TIMESTAMP))
  AND g_dnsa_dt_yyyymmdd <> CAST('9999-12-31' AS TIMESTAMP);

/* Validate g_expiration_dt_yyyymmdd is valid or default */
SELECT
    COUNT(*) AS invalid_expiration_dt
FROM purgo_playground.f_invntry_bal_dly_hist
WHERE (g_expiration_dt_yyyymmdd NOT BETWEEN CAST('0001-01-01' AS TIMESTAMP) AND CAST('9999-12-31' AS TIMESTAMP))
  AND g_expiration_dt_yyyymmdd <> CAST('9999-12-31' AS TIMESTAMP);
```
